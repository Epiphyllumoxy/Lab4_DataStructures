```python
# My Inputs
LAST_NAME = "CULATA"         

customers = [f"{LAST_NAME}_C{i}" for i in range(1, 6)]
products = ["NOTEBOOK", "PEN", "USB", "HEADPHONES", "CALCULATOR"]
prices = {"NOTEBOOK": 3.50, "PEN": 1.20, "USB": 15.00, "HEADPHONES": 25.00, "CALCULATOR": 18.50}

# 1. Generated Transaction Records Input Block
raw_transactions = [
    {"customer": customers[0], "product": products[0], "quantity": 3},
    {"customer": customers[1], "product": products[1], "quantity": 5},
    {"customer": customers[2], "product": products[2], "quantity": 1},
    {"customer": customers[3], "product": products[3], "quantity": 2},
    {"customer": customers[4], "product": products[4], "quantity": 1},
    {"customer": customers[0], "product": products[0], "quantity": 3}  # Repeated Purchase
]

# 2, 3, 4, 5. Tabulate pricing matrices across mapped keys
customer_spend = {}
product_revenue = {}
total_store_revenue = 0.0

for trans in raw_transactions:
    cust, prod, qty = trans["customer"], trans["product"], trans["quantity"]
    item_cost = prices[prod] * qty
    
    total_store_revenue += item_cost
    customer_spend[cust] = customer_spend.get(cust, 0.0) + item_cost
    product_revenue[prod] = product_revenue.get(prod, 0.0) + item_cost

# 6 & 7. Order results dynamically based on revenue accumulation criteria
ordered_customers = sorted(customer_spend.items(), key=lambda x: x[1], reverse=True)

print("\n" + "=" * 72)
print("                    FINAL CONSOLIDATED TRANSACTION REPORT               ")
print("=" * 72)
print(f"{'Customer Code':<20}{'Gross Expenditure'}")
print("-" * 72)
for cust, spend in ordered_customers:
    print(f"{cust:<20}${spend:.2f}")
print("-" * 72)
print(f"{'TOTAL RETAIL STORE REVENUE TRAFFIC':<20}${total_store_revenue:.2f}")
print("=" * 72)
```
