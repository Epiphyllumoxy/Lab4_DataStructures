```python
# My Inputs
LAST_NAME = "CULATA"         
SEED_NUM = 8                
FAVORITE_ARTIST = "ARIANA GRANDE"  

# Helper to derive numerical generation seed bounds
char_sum = sum(ord(char.upper()) - 64 for char in FAVORITE_ARTIST if char.isalpha())
base_val = SEED_NUM + (char_sum % 50)

# Rules-driven arrays setup
sources = [f"{LAST_NAME}_T{i}" for i in range(1, 6)]
categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]

# 1. Generated Test Data List
raw_test_data = [
    {"source": sources[0], "category": categories[0], "value": round(base_val * 1.5, 1)},
    {"source": sources[1], "category": categories[1], "value": round(base_val * 0.2, 1)},
    {"source": sources[2], "category": categories[2], "value": round(base_val * 0.1, 1)},
    {"source": sources[3], "category": categories[3], "value": round(base_val * 2.1, 1)},
    {"source": sources[4], "category": categories[4], "value": round(base_val * 2.5, 1)},
    {"source": sources[0], "category": categories[0], "value": round(base_val * 1.5, 1)},  # Repeated Record
    {"source": sources[1], "category": categories[1], "value": round(base_val * 0.3, 1)},  # Additional Data
    {"source": sources[2], "category": categories[2], "value": round(base_val * 0.15, 1)}  # Additional Data
]

print("1. Generated Test Data:")
for entry in raw_test_data:
    print(f"   {entry}")

# 2. Organize Test Data into dictionary map mapping
organized_data = {}
for entry in raw_test_data:
    src, cat, val = entry["source"], entry["category"], entry["value"]
    if src not in organized_data:
        organized_data[src] = {}
    if cat not in organized_data[src]:
        organized_data[src][cat] = []
    organized_data[src][cat].append(val)

# 3. Identify Repeated and Distinct Records
seen_records, repeated_records = [], []
for entry in raw_test_data:
    record_tuple = (entry["source"], entry["category"], entry["value"])
    if record_tuple in seen_records:
        if record_tuple not in repeated_records:
            repeated_records.append(record_tuple)
    else:
        seen_records.append(record_tuple)

print(f"\n3. Repeated Records Found: {repeated_records}")
print(f"   Distinct Records Count: {len(seen_records)}")

# 4. Determine numerical and categorical summaries
print("\n4. Numerical and Categorical Summaries:")
cat_summaries = {}
for entry in raw_test_data:
    cat = entry["category"]
    if cat not in cat_summaries:
        cat_summaries[cat] = []
    cat_summaries[cat].append(entry["value"])

for cat, vals in cat_summaries.items():
    avg_v = sum(vals) / len(vals)
    print(f"   - {cat:<12} : Count={len(vals)}, Mean={avg_v:.2f}, Min={min(vals)}, Max={max(vals)}")

# 5, 6, 7. Final Organized Output Display
print("\n" + "=" * 72)
print("                 FINAL ORGANIZED AND SUMMARIZED TEST DATASET            ")
print("=" * 72)
for src in sorted(organized_data.keys()):
    print(f"Source: {src}")
    for cat in sorted(organized_data[src].keys()):
        vals = organized_data[src][cat]
        avg_v = sum(vals) / len(vals)
        print(f"  - {cat:<12} : {vals} (Avg: {avg_v:.2f}, Min: {min(vals):.2f}, Max: {max(vals):.2f})")
print("=" * 72)
```
